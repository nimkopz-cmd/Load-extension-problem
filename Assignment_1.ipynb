```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv("Load_Extension.csv")
print(data)
```

```python
x=data[["Load (N)"]]
y=data[["Extension (mm)"]]
plt.scatter(x,y)
plt.xlabel("Load")
plt.ylabel("Extension")
plt.title("Load vs Extension")
plt.show()
model=LinearRegression()
model.fit(x,y)
inp_user=int(input("Enter a Load (N):"))
new_extension=model.predict([[inp_user]])
print("Extension is",new_extension,"mm")
```
